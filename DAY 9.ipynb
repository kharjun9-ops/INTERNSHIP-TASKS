```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Department': ['HR', 'IT', 'Finance', 'Marketing', 'Sales'],
    'Employees': [15, 45, 30, 22, 50],
    'Budget_USD': [50000, 150000, 120000, 80000, 200000],
    'Q1_Revenue': [12000, 25000, 18000, 15000, 30000],
    'Q2_Revenue': [15000, 28000, 20000, 18000, 35000]
}
df = pd.DataFrame(data)

print("==========================================")
print("          BUSINESS SUMMARY METRICS        ")
print("==========================================")
print(f"Total Employees:    {df['Employees'].sum()}")
print(f"Total Budget:       ${df['Budget_USD'].sum():,.2f}")
print(f"Average Department Budget: ${df['Budget_USD'].mean():,.2f}")
print("==========================================")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Departmental Performance Dashboard', fontsize=16, fontweight='bold')

axes[0, 0].bar(df['Department'], df['Employees'], color='skyblue', edgecolor='black')
axes[0, 0].set_title('Employee Count by Department')
axes[0, 0].set_ylabel('Number of Staff')
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.7)

axes[0, 1].pie(df['Budget_USD'], labels=df['Department'], autopct='%1.1f%%', startangle=140)
axes[0, 1].set_title('Budget Allocation (%)')

axes[1, 0].plot(df['Department'], df['Q1_Revenue'], marker='o', label='Q1 Revenue', color='blue')
axes[1, 0].plot(df['Department'], df['Q2_Revenue'], marker='s', label='Q2 Revenue', color='green')
axes[1, 0].set_title('Quarterly Revenue Growth (Q1 vs Q2)')
axes[1, 0].set_ylabel('Revenue ($)')
axes[1, 0].legend()
axes[1, 0].grid(True)

axes[1, 1].barh(df['Department'], df['Budget_USD'], color='orange', alpha=0.8)
axes[1, 1].set_title('Total Budget by Department')
axes[1, 1].set_xlabel('Budget ($)')
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
